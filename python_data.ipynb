```python
# requests library is used to make HTTP requests

import requests
```

```python
url = "http://api.weatherapi.com/v1/current.json?key=7809c64d700d4c8cb86162822260205&q=Delhi&aqi=yes"

x = requests.get(url)
print(x.status_code)
print(x.text)

```

```python
data = x.json()
print(data)
```

```python
city  = data['location']['name']
print(city)
```

```python
details = data['current']['condition']['text']
print(details)
```

```python
air = data['current']['air_quality']
print(air)
```

```python
city = data["location"]["name"]
region = data["location"]["region"]

temp = data["current"]["temp_c"]
humidity = data["current"]["humidity"]
wind_speed = data["current"]["wind_kph"]

condition = data["current"]["condition"]["text"]
air_quality = data["current"]["air_quality"]["pm2_5"]

print(f"City: {city}, Region: {region}")
print(f"Temperature: {temp}°C, Humidity: {humidity}%, Wind Speed: {wind_speed} kph")
print(f"Condition: {condition}, Air Quality (PM2.5): {air_quality}")    
```

```python
# make it in a dictonary like foramt
weather_data = {
    "city": city,
    "region": region,
    "temperature_c": temp,
    "humidity": humidity,
    "wind_speed_kph": wind_speed,
    "condition": condition,
    "air_quality_pm2_5": air_quality
}
```

```python
import pandas as pd

df = pd.DataFrame([weather_data])

print(df)
```

```python
# save it to csv format and it rewite the original one
df.to_csv("Ranchi_weather.csv", index = False)
```

```python
# this will append every new data to the existing csv file without overwriting it
df.to_csv("Ranchi_weather_clean.csv", mode='a', header=False, index=False)
```

```python
cities = ["Delhi", "Mumbai", "Bangalore", "Kolkata"]
```

```python
import requests

all_data = []

for city in cities:
    url = f"http://api.weatherapi.com/v1/current.json?key=7809c64d700d4c8cb86162822260205&q={city}&aqi=yes"
    
    response = requests.get(url)
    data = response.json()

    clean_data = {
        "city": data["location"]["name"],
        "temp": data["current"]["temp_c"],
        "humidity": data["current"]["humidity"],
        "condition": data["current"]["condition"]["text"]
    }

    all_data.append(clean_data)
    
print(all_data)
```

```python
import pandas as pd

df = pd.DataFrame(all_data)

print(df)
```

```python
df.to_csv("multiple_cities_weather.csv",index=False)
```
