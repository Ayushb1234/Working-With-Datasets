```python

# Homework Question 1: Employee Salary Calculation
# ----------------------------------------------------

employee_id = int(input("Enter Employee ID: "))

name = input("Enter Employee Name: ")

salary = float(input("Enter Salary: "))

exp = int(input("Enter Experience (in years): "))

if exp >= 5:
    bonus = salary * 0.05   
else:
    bonus = salary * 0.02

total_salary = salary + bonus

print(" Employee Details")

print("Employee ID :", employee_id)

print("Employee Name :", name)

print("Salary :", salary)

print("Experience :", exp, "years")

print("Bonus :", bonus)

print("Total Salary :", total_salary)
```

```python
import numpy as np

n = int(input("Enter number of employees: "))

emp_dtype = np.dtype([
    ('emp_id', 'i4'),
    ('name', 'U20'),
    ('department', 'U20'),
    ('salary', 'f8'),
    ('experience', 'i4')
])

employees = np.empty(n, dtype=emp_dtype)

for i in range(n):
    print(f"\nEnter details of Employee {i+1}:")

    employees[i]['emp_id'] = int(input("Employee ID: "))
    employees[i]['name'] = input("Name: ")
    employees[i]['department'] = input("Department: ")
    employees[i]['salary'] = float(input("Salary: "))
    employees[i]['experience'] = int(input("Experience (years): "))

print("\n------ Employee Details ------")

for emp in employees:
    print("\nEmployee ID :", emp['emp_id'])
    print("Name        :", emp['name'])
    print("Department  :", emp['department'])
    print("Salary      :", emp['salary'])
    print("Experience  :", emp['experience'], "years")

print("\n------ Bonus Calculation Based on Experience ------")

for i in range(n):

    if employees[i]['experience'] > 5:

        bonus = employees[i]['salary'] * 0.20
        employees[i]['salary'] += bonus

        print(f"{employees[i]['name']} receives bonus of {bonus:.2f}")

    else:
        print(f"{employees[i]['name']} does not receive bonus")

print("\n------ Final Salary After Bonus ------")

for emp in employees:
    print(f"{emp['name']} -> Final Salary: {emp['salary']:.2f}")

total_payment = np.sum(employees['salary'])

print(f"\nTotal Payment After Bonus = {total_payment:.2f}")
```
