```python
# Task 5:
# -------------

# List Analytics System
# -------------------------

ans = []

for i in range(5):
    
    x = int(input("ENter values:"))
    ans.append(x)
    
print(ans)

print(max(ans))
print(min(ans))

sum = 0

for i in range(len(ans)):
    
    sum += ans[i]
    
print(sum)  

print("Average: ", sum / len(ans))

print("-----------------------------------")

for x in ans:
    if x % 2 == 0:
        print(x)

print("-----------------------------------")


for x in ans:
    if x % 2 != 0:
        print(x)




  
```

```python
while True:
    
    print("\n------------- MENU ----------------")
    print("1. ADD Notes")
    print("2. View Notes")
    print("3. Exit")

    inp = input("Enter your choice: ")

    # ADD NOTES
    if inp == "1":

        with open("Notes.txt", "a") as notes:

            etr = input("Add Note: ")
            notes.write(etr + "\n")

        print("Note Added Successfully!")

    # VIEW NOTES
    elif inp == "2":

        with open("Notes.txt", "r") as notes:

            content = notes.readlines()

            print("\n------ YOUR NOTES ------")

            for note in content:
                print(note, end="")

    # EXIT
    elif inp == "3":

        print("Exiting Notes App...")
        break

    # INVALID INPUT
    else:
        print("Invalid Choice!")
```

```python
# create a ATM system
# -----------------------
balance = 10000

while(True):
    
    print("----------------- Options --------------------")
    print("1. Withdraw")
    print("2. Deposit")
    print("3. Check balance")
    print("4. Exit")
    
    inp = int(input("Enter your choice:"))
    
    
    if inp == 1:
        x = int(input("Enter amount to withdraw: "))
        
        if balance == 0 & x > balance:
            print("no withdraawl available")
            break
        else:
            balance = balance - x
            print(f"balance remaining :{balance}")
            print(f"Amount withdrawn :{x}")
    
    elif inp == 2:
        y = int(input("Enter amount to Deposit: "))
        
        balance  = balance + y
        
        print(f"amount deposited:{y}")
        print(f"updated balance:{balance}")
        
    elif inp == 3:
        
        print(f"current balance: {balance}")
        
    elif inp == 4:
        print("Exit successfully")
        break
    
    else:
        print("Invalid Option")
        break
    
    
    
    
    
```
