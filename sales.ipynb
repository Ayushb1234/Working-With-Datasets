```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Salesperson_Name": ["Rahul", "Priya", "Aman", "Sneha", "Karan"],
    "Region": ["North", "South", "East", "West", "Central"],
    "Product_Category": ["Laptop", "Mobile", "Furniture", "Tablet", "Accessories"],
    "Monthly_Sales": [250000, 180000, 320000, 210000, 150000],
    "Target_Achieved_%": [95, 88, 110, 92, 75],
    "Years_of_Experience": [2, 4, 6, 3, 1]
}
```

```python
df = pd.DataFrame(sales_data)
print(df)
```

```python
# Data frame properties
# Data Operations
# Filtering
# Adding Coulmns in Dataframe
# Updating the Dataframe
# Missing Values
# Sorting of the Data
# groupby Function Apply 
# Finally Dropping and saving the data
# matplotlib plotting visuals also
```

```python
# Data frame properties
print(df.shape)
print("--------------------------------------------\n")
print(df.info())
print("--------------------------------------------\n")
print(df.describe())
print("--------------------------------------------\n")
print(df.columns)
print("--------------------------------------------\n")
print(df.dtypes)
print("--------------------------------------------\n")
print(df.head())
print("--------------------------------------------\n")
print(df.index)

```

```python
# Data Operations
# --------------------
print(df["Monthly_Sales"])
print("--------------------------------------------\n")
print(df[["Salesperson_Name","Product_Category","Years_of_Experience"]])
print("--------------------------------------------\n")
print(df.loc[2])
print("--------------------------------------------\n")
print(df.iloc[3])
print("--------------------------------------------\n")
print(df.loc[1:3,["Salesperson_Name","Monthly_Sales"]])
```

```python
# Filtering rows and columns
# ------------------------------

high_sales = df[df["Monthly_Sales"] > 200000]
print(high_sales)
print("----------------------------------------------------------------------------\n")

highest_exp = df[df["Years_of_Experience"] > 3]
print(highest_exp)
print("---------------------------------------------------------------------------\n")

exp_employee = df[(df["Years_of_Experience"] > 3) & (df["Monthly_Sales"] > 250000)]
print(exp_employee)
```

```python
# Adding new columns in the DataFrame
# -----------------------------------------

df["Sales_target"] = df["Monthly_Sales"] * 1.05 # 5 % increase in sales target
print(df)
```

```python
# Updating the dataframe
# -----------------------------

df.loc[2,"Monthly_Sales"] = 450000
print(df)
```

```python
df.loc[df["Employee_ID"] == 105, "Target_Achieved_%"] = 95
print(df)
```

```python
# Missing Values Operation
# --------------------------

df.loc[df[2:5].index,"Monthly_Sales"] = np.nan
df.loc[df[2:4].index,"Target_Achieved_%"] = np.nan
df.loc[df[1:4].index,"Sales_target"] = np.nan
print(df)

```

```python
df.isna().sum()
```

```python
df["Region"].value_counts()
```

```python
df.isna().sum()/len(df) * 100
```

```python
df["Monthly_Sales"].fillna(df["Monthly_Sales"].mean(), inplace = True)
print(df)
print("--------------------------------------------\n")

df["Target_Achieved_%"].fillna(df["Target_Achieved_%"].mean(), inplace = True)
print(df)
print("--------------------------------------------\n")

df["Sales_target"].fillna(df["Sales_target"].mean(), inplace = True)
print(df)
print("--------------------------------------------\n")
```

```python
df["Monthly_Sales"] = df["Monthly_Sales"].fillna(df["Monthly_Sales"].mean())

df["Target_Achieved_%"] = df["Target_Achieved_%"].fillna(
    df["Target_Achieved_%"].mean()
)

df["Sales_target"] = df["Sales_target"].fillna(
    df["Sales_target"].mean()
)

print(df)
```

```python
# Sorting Data Operations
# ------------------------------

sorted_values_1 = df.sort_values(by = "Salesperson_Name", ascending = True)

sorted_values_2 = df.sort_values(by = "Target_Achieved_%",ascending  = False)

print(sorted_values_1)

print("----------------------------------------------------------------------------\n")

print(sorted_values_2)

```

```python
# Groupby Function Apply
# ---------------------------

avg_sales_region = df.groupby("Region")["Monthly_Sales"].mean()
print(avg_sales_region)


print("---------------------------------------------------------------\n")

person_age = df.groupby("Salesperson_Name")["Years_of_Experience"].mean()
print(person_age)


```

```python
# Saving the Data
# ----------------

df.to_csv("sales_dataset.csv", index = False)
```

```python
# # Graphical Representation
# # -------------------------------


# Q. which region has maximum sales?

df[["Region","Monthly_Sales"]].groupby("Region").sum().plot(kind = "bar")
plt.title("Total Sales By region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()



# Q. which product has maximum sales?

df.groupby("Product_Category")["Monthly_Sales"].sum().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Total Sales By Product")
plt.ylabel("")
plt.show()

# Q. which employee has maximum sales?

df.groupby("Salesperson_Name")["Monthly_Sales"].sum().plot(
    kind = "line",
    marker = "o",
    linestyle = ":",
    color = "red",
)

plt.title("Total sales by Employee")
plt.xlabel("Employee Name")
plt.ylabel("Total Sales")
plt.xticks(rotation = 45)
plt.legend()
plt.grid()
plt.show()
```
