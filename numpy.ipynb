```python
range = np.arange(1,30,3)
print(range)
# what is thiss range 
```

```python
arr= np.linspace(0,1,5)
print(arr)
# it divides the range into 5 equal parts and gives us the values in those parts.
```

```python
arr = np.logspace(2,3,5)
print(arr)
# it divides the range into 5 equal parts and gives us the values in those parts but in log scale.
```
