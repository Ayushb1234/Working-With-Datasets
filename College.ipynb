```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    "College_ID": [1, 2, 3, 4, 5],
    "Name":["Ayush","Satya","Jiban","Ritesh","Santosh"],
    "Department":["CSE","ECE","ME","Civil","Architecture"],
    "Fees":[50000, 45000, 40000, 55000, 60000],
    "Current_year":[2,3,4,1,4]
}

df = pd.DataFrame(data)
print(df)
```

```python
# data frame properties
# --------------------------

print(df.shape)
print("--------------------")
print(df.columns)
print("--------------------")

print(df.dtypes)
print("--------------------")

print(df.index)
```

```python
# Data Operations
# ------------------
print(df["Fees"])
print("------------------------------------\n")
print(df[["Name","Department","Fees"]])

print("------------------------------------\n")
print(df.loc[0]) # index based location
print("------------------------------------\n")
print(df.iloc[3]) # position based loction prints  4th row details
print("------------------------------------\n")
print(df.loc[3,"Fees"]) # index at 3 shows fees
```

```python
# Filtering
# --------------

Highest_Fees = df["Fees"] > 30000
print(df[Highest_Fees])
```

```python
Seniority = df[(df["Current_year"] > 2) & (df["Fees"] > 45000)]
print(Seniority)
```

```python
#  Adding Coulms in Dataframe
# -------------------------------------

df["Scholarship"] = df["Fees"] > 45000 # returns true if condition satisfies else false
print(df)
```

```python
# Updating the Data
# ----------------------

df.loc[2,"Fees"] = 40000
print(df)
```

```python
# Missing Values
# ---------------------
df.loc[2,"Current_year"] = np.nan
print(df)
```

```python
print(df.isna().sum())
```

```python
print(df.isna())
```

```python
print(df.isna().sum()/len(df) * 100) # tells the percentage of missing values in cols
```

```python
df["Current_year"] = df["Current_year"].fillna(df["Current_year"].mean())

print(df)
```

```python
# Sorting of the Data
# ---------------------

sorted_Fees = df.sort_values(by="Fees",ascending=False)
print(sorted_Fees)
```

```python
# groupby Function Apply 
# ----------------------------

avg_Fees = df.groupby("Department")["Fees"].mean()
print(avg_Fees)
```

```python
# Finally Dropping and saving the data
# ----------------------------------------

df.drop("Scholarship",axis = 1,inplace= True)
print(df)
```

```python
df.drop(3,axis = 0,inplace = True)
print(df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# loading the dataset from seaborn library
data = sns.load_dataset("tips")

print("First 5 rows of Dataset:")
print(data.head())   # prints first 5 rows

# if you use head it prints first rows
# tail() prints last rows

print("\nDataset Information:")
print(data.info())   # shows dtype, column names, memory usage etc.

sns.set_style("whitegrid")

plt.figure(figsize=(8,5))

sns.lineplot(x="size", y="total_bill", data=data)

plt.title("Total Bill vs Group Size")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(x="size", y="total_bill", data=data)

plt.title("Distribution of Total Bill")

plt.xlabel("Group Size")

plt.ylabel("Total Bill")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data["total_bill"], bins=10)
# bins = the dist of our data in bars

plt.title("dist of total bill")

plt.show()


plt.figure(figsize=(8,5))

sns.violinplot(x="sex", y="total_bill", data=data)

plt.title("bill paid by particular gender")

plt.show()


plt.figure(figsize=(8,5))

sns.barplot(x="day", y="time", data=data)

plt.title("customer per day")

plt.show()


sns.pairplot(data)

plt.show()
```
