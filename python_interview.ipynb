```python
#  programm to show case all oops concepts

class animal:
    def __init__(self,name):
        self.name = name
        self.species = "unknown"
    
    def make_sound(self):
        print("animal sound")
        
class dog(animal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed = breed
        self.species = "dog"
        
    def make_sound(self):
        print("bark")
        
```

```python
# What is GIL? - It stands for global interpreter lock. here one thread can execute at a time  to avoid race conditions.

# Why Python slower than C? - Python is an interpreted language while C is a compiled language. Python code is executed line by line which adds overhead, while C code is compiled into machine code that runs directly on the hardware, making it faster.

# List comprehension -  concise way for creating lists, it is more easier and faster than the normal way of creating loops.

# Generators vs normal functions - genratiers is used to generate a sequence of values one at a time, while normal functions return a single value. generators are more memory efficient as they do not store the entire sequence in memory.


# Iterator - follows the iterator protocol which consist of __iter__() and __next__() methods. it allows you to iterate over a collection of items one at a time. it is used in for loops and other iteration contexts.

# simple example:

a = [1,2,3]

it = iter(a)
print(next(it))
print(next(it))

# genrator

def count(n):
    i = 0
    while i < n:
        yield i
        i += 1

print(list(count(10)))
    
```

```python
import numpy as np
import copy

original = np.array([[1, 2], [3, 4], [5, 6]])

print(original.ndim)

shallow = copy.copy(original)
deep = copy.deepcopy(original)

print("Original:", original)
print("Shallow:", shallow)
print("Deep:", deep)

# then what is differnce in outut they all same 
# then do changes
original[0][0] = 10
# shallow[0][0] = 9
print("Original:", original)
print("Shallow:", shallow)
print("Deep:", deep)

# that means we can say that in changung the values in orginal other two also changes but when we change the value in deep copy it does not change the value in original and shallow copy.
```
